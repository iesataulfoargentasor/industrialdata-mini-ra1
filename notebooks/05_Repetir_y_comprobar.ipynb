Repetir una recepción debe quedar registrado, pero no duplicar observaciones ni episodios. Las pruebas usan casos pequeños independientes; no leen el listado de escenarios para decidir qué detectar.

```python
import json
import pandas as pd
from industrialdata.common import root
from industrialdata.run import execute, current_snapshot
folder = root()
if current_snapshot(folder) is None:
    execute(folder)
before = pd.read_parquet(current_snapshot(folder) / "gold/metrics.parquet")
report = execute(folder)
after = pd.read_parquet(current_snapshot(folder) / "gold/metrics.parquet")
assert report["accepted_new"] == 0 and report["duplicates"] == 41 and report["rejected"] == 4
assert report["canonical_rows"] == 40 and report["episodes"] == 1
pd.testing.assert_frame_equal(before, after)
print("Segunda carga: 45 = 0 nuevas + 41 duplicadas + 4 rechazadas. Consolidadas: 40.")
```

## Una prueba pequeña que puedes explicar

```python
from industrialdata.common import timestamp
from industrialdata.reglas import energia
from datetime import timedelta
a = timestamp("2026-01-12T08:00:00Z")
times = [a + timedelta(minutes=i) for i in range(6)]
kwh, seconds = energia(times, [6]*6, a, a + timedelta(minutes=5))
assert abs(kwh - .5) < 1e-9 and seconds == 300
print("6 kW durante 5 minutos: 0,5 kWh; 300 segundos cubiertos.")
```
