# 04 · Publicar y mostrar resultados (62–72 min)

Esta celda ejecuta el mismo pipeline que se puede lanzar desde Terminal. Guarda Silver y Gold en Parquet, rechazos con procedencia y un informe. El panel solo ve una instantánea cuando está completa. Los originales no cambian. Si ya ejecutaste la carga, el informe mostrará cero nuevas y 41 duplicadas: es correcto.

```python
import json
import pandas as pd
from industrialdata.common import root, sha
from industrialdata.run import execute, current_snapshot
folder = root()
raw = folder / "raw/telemetry_001.jsonl"
before = sha(raw)
report = execute(folder)  # Catálogo y registro de cargas en PostgreSQL.
assert report["catalog_source"] == "PostgreSQL"
assert sha(raw) == before
print(json.dumps(report, indent=2))
```

```python
snapshot = current_snapshot(folder)
print("Publicación:", snapshot)
display(pd.read_parquet(snapshot / "gold/metrics.parquet"))
assert len(pd.read_parquet(snapshot / "silver/telemetry.parquet")) == 40
```
